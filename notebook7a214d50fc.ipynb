```python
# ============================================================
# Brain Tumor MRI Dataset - 3 Modern Models + Visualizations
# Dataset: https://www.kaggle.com/datasets/masoudnickparvar/brain-tumor-mri-dataset
# Models:
#   1) Vision Transformer (ViT) - custom, fine-tuned
#   2) Xception (Transfer Learning)
#   3) Improved EfficientNetB0 (Custom EfficientNet variant)
# EPOCHS = 100 (with EarlyStopping)
# + Data visualization for each tumor type
# ============================================================

# ----------------- Basic libs & data path detection -----------------
import numpy as np
import pandas as pd
import os
import random

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

print("Listing /kaggle/input contents:")
for dirname, _, _ in os.walk('/kaggle/input'):
    print(dirname)

# Auto-detect dataset folder under /kaggle/input
data_path = None
for dirname, _, _ in os.walk('/kaggle/input'):
    if 'brain-tumor-mri-dataset' in dirname:
        data_path = dirname
        break

if data_path is None:
    raise FileNotFoundError("Could not find 'brain-tumor-mri-dataset' under /kaggle/input")

print(f"\nUsing data_path: {data_path}")

train_dir = os.path.join(data_path, "Training")
test_dir  = os.path.join(data_path, "Testing")
print("Train dir:", train_dir)
print("Test dir :", test_dir)

# ----------------- TensorFlow & Keras imports -----------------
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import Xception, EfficientNetB0

print("TensorFlow version:", tf.__version__)

# ----------------- Global config -----------------
IMG_SIZE   = (224, 224)
BATCH_SIZE = 32
SEED       = 42
EPOCHS     = 100   # as requested

# ============================================================
# DATA PIPELINE (TRAIN / VALIDATION / TEST)
# ============================================================
train_datagen = ImageDataGenerator(
    rescale=1.0/255,
    rotation_range=15,
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.1,
    horizontal_flip=True,
    validation_split=0.2,
)

test_datagen = ImageDataGenerator(rescale=1.0/255)

train_gen = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="training",
    seed=SEED
)

val_gen = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="validation",
    seed=SEED
)

test_gen = test_datagen.flow_from_directory(
    test_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

class_indices = train_gen.class_indices
num_classes = len(class_indices)
print("Classes:", class_indices)

idx_to_class = {v: k for k, v in class_indices.items()}
ordered_class_names = [idx_to_class[i] for i in range(len(idx_to_class))]

# ============================================================
# DATA VISUALIZATION FOR EACH TUMOR TYPE
#   1) Class distribution (train/val/test)
#   2) Sample images per class (from Training set)
# ============================================================

def plot_class_distribution(generator, title, ordered_class_names):
    labels = generator.classes
    counts = np.bincount(labels, minlength=len(ordered_class_names))

    plt.figure(figsize=(8, 5))
    plt.bar(ordered_class_names, counts)
    plt.title(title)
    plt.xlabel("Class (Tumor Type)")
    plt.ylabel("Number of images")
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.show()

    for name, c in zip(ordered_class_names, counts):
        print(f"{title} - {name}: {c} images")

print("\n--- Class Distribution (Number of images per tumor type) ---")
plot_class_distribution(train_gen, "Training Set Distribution",   ordered_class_names)
plot_class_distribution(val_gen,   "Validation Set Distribution", ordered_class_names)
plot_class_distribution(test_gen,  "Test Set Distribution",       ordered_class_names)

def show_sample_images_per_class(base_dir, class_names, n_samples=4):
    """
    Show a grid of sample images for each class (tumor type).
    Rows    -> classes
    Columns -> samples per class
    Tumor name is written in front of (above) each image.
    """
    def pretty_name(name):
        # "glioma_tumor" -> "Glioma Tumor", "no_tumor" -> "No Tumor"
        return name.replace("_", " ").title()

    n_classes = len(class_names)
    plt.figure(figsize=(n_samples * 3, n_classes * 3))

    for row, cls_name in enumerate(class_names):
        class_dir = os.path.join(base_dir, cls_name)
        img_files = [
            f for f in os.listdir(class_dir)
            if f.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"))
        ]

        if len(img_files) == 0:
            continue

        random.shuffle(img_files)
        use_n = min(n_samples, len(img_files))

        for col in range(use_n):
            idx = row * n_samples + col + 1
            plt.subplot(n_classes, n_samples, idx)
            img_path = os.path.join(class_dir, img_files[col])
            img = plt.imread(img_path)

            if img.ndim == 2:
                plt.imshow(img, cmap="gray")
            else:
                plt.imshow(img)

            plt.axis("off")
            # tumor name above each image
            plt.title(pretty_name(cls_name), fontsize=10)

    plt.suptitle("Sample Training Images for Each Tumor Type", fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

print("\n--- Sample images from each tumor type (Training set) ---")
show_sample_images_per_class(train_dir, ordered_class_names, n_samples=4)

# ============================================================
# HELPER: TRANSFORMER ENCODER BLOCK (for ViT)
# ============================================================
def transformer_encoder(inputs, num_heads, mlp_dim, dropout=0.1):
    # Layer norm 1 + Multi-head Self-Attention
    x = layers.LayerNormalization(epsilon=1e-6)(inputs)
    x = layers.MultiHeadAttention(
        num_heads=num_heads,
        key_dim=inputs.shape[-1],
        dropout=dropout
    )(x, x)
    x = layers.Dropout(dropout)(x)
    x = layers.Add()([x, inputs])  # residual 1

    # Layer norm 2 + MLP
    y = layers.LayerNormalization(epsilon=1e-6)(x)
    y = layers.Dense(mlp_dim, activation="gelu")(y)
    y = layers.Dropout(dropout)(y)
    y = layers.Dense(inputs.shape[-1])(y)
    y = layers.Dropout(dropout)(y)

    # Residual 2
    return layers.Add()([x, y])

# ============================================================
# MODEL 1: Vision Transformer (ViT) - Custom Implementation
# ============================================================
def build_vit_model(
    image_size=224,
    patch_size=16,
    num_layers=8,
    d_model=256,
    num_heads=8,
    mlp_dim=512,
    dropout=0.1,
    num_classes=4,
):
    """
    ViT-style model without explicit class token:
    - Patch embedding via Conv2D
    - Positional embeddings via Embedding
    - Stacked Transformer encoder blocks
    - GlobalAveragePooling over tokens for classification
    """

    # Input
    inputs = keras.Input(shape=(image_size, image_size, 3))

    # Patch embedding: Conv2D with stride=patch_size
    x = layers.Conv2D(
        filters=d_model,
        kernel_size=patch_size,
        strides=patch_size,
        padding="valid",
        name="patch_embedding"
    )(inputs)

    # Flatten spatial dims: (batch, H_p, W_p, d_model) -> (batch, N_patches, d_model)
    patch_dims = (image_size // patch_size) ** 2
    x = layers.Reshape((patch_dims, d_model))(x)  # (batch, N, d_model)

    # Positional embeddings
    positions = tf.range(start=0, limit=patch_dims, delta=1)  # (N,)
    pos_embed_layer = layers.Embedding(
        input_dim=patch_dims,
        output_dim=d_model,
        name="positional_embedding"
    )
    pos_embeddings = pos_embed_layer(positions)  # (N, d_model)

    # Add position info
    x = x + pos_embeddings

    # Transformer encoder blocks
    for _ in range(num_layers):
        x = transformer_encoder(
            x,
            num_heads=num_heads,
            mlp_dim=mlp_dim,
            dropout=dropout
        )

    # Final norm + pooling over tokens
    x = layers.LayerNormalization(epsilon=1e-6)(x)
    x = layers.GlobalAveragePooling1D()(x)  # (batch, d_model)

    # Classification head
    x = layers.Dropout(dropout)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = keras.Model(inputs=inputs, outputs=outputs, name="ViT_brain_tumor")
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-4),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

# ============================================================
# MODEL 2: Xception (Transfer Learning)
# ============================================================
def build_xception_model(num_classes=4, img_size=(224, 224)):
    base_model = Xception(
        include_top=False,
        weights="imagenet",
        input_shape=img_size + (3,)
    )
    base_model.trainable = False  # frozen feature extractor

    inputs = keras.Input(shape=img_size + (3,))
    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(512, activation="relu")(x)
    x = layers.Dropout(0.4)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = keras.Model(inputs, outputs, name="Xception_brain_tumor")
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-4),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

# ============================================================
# MODEL 3: Improved EfficientNetB0 (Custom Variant)
# ============================================================
def build_improved_efficientnet(num_classes=4, img_size=(224, 224)):
    """
    'Improved' EfficientNet idea:
    - EfficientNetB0 backbone (pretrained)
    - Deeper custom classification head with Dropout & BN
    """
    base_model = EfficientNetB0(
        include_top=False,
        weights="imagenet",
        input_shape=img_size + (3,)
    )
    base_model.trainable = False  # start frozen

    inputs = keras.Input(shape=img_size + (3,))
    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)

    # Custom "improved" head
    x = layers.BatchNormalization()(x)
    x = layers.Dense(512, activation="relu")(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = keras.Model(inputs, outputs, name="Improved_EfficientNetB0")
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-4),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

# ============================================================
# HELPER: EVALUATION + CONFUSION MATRIX / REPORT
# ============================================================
def evaluate_model_on_test(model, test_gen, class_indices, model_name):
    """
    - Evaluates model on test set
    - Computes confusion matrix
    - Prints classification report
    """
    # Evaluate -> loss & accuracy
    test_gen.reset()
    test_loss, test_acc = model.evaluate(test_gen, verbose=1)

    # Predictions for confusion matrix
    test_gen.reset()
    y_pred_prob = model.predict(test_gen, verbose=1)
    y_pred = np.argmax(y_pred_prob, axis=1)
    y_true = test_gen.classes

    idx_to_class_local = {v: k for k, v in class_indices.items()}
    class_names = [idx_to_class_local[i] for i in range(len(idx_to_class_local))]

    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names
    )
    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    plt.title(f"{model_name} - Confusion Matrix (Test)")
    plt.tight_layout()
    plt.show()

    print(f"\nClassification Report for {model_name} (per tumor type):\n")
    print(classification_report(y_true, y_pred, target_names=class_names))

    return test_loss, test_acc

# ============================================================
# TRAINING LOOP FOR ALL THREE MODELS
# ============================================================
callbacks_common = [
    keras.callbacks.EarlyStopping(
        monitor="val_accuracy",
        patience=10,
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-6,
        verbose=1
    )
]

models_builders = {
    "ViT": lambda: build_vit_model(
        image_size=IMG_SIZE[0],
        patch_size=16,
        num_layers=8,
        d_model=256,
        num_heads=8,
        mlp_dim=512,
        dropout=0.1,
        num_classes=num_classes
    ),
    "Xception": lambda: build_xception_model(
        num_classes=num_classes,
        img_size=IMG_SIZE
    ),
    "Improved_EfficientNet": lambda: build_improved_efficientnet(
        num_classes=num_classes,
        img_size=IMG_SIZE
    )
}

histories = {}
results   = {}

for name, builder in models_builders.items():
    print("\n" + "="*70)
    print(f"Training model: {name}")
    print("="*70)

    tf.keras.backend.clear_session()
    model = builder()
    model.summary()

    history = model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=EPOCHS,
        callbacks=callbacks_common,
        verbose=1
    )
    histories[name] = history.history

    print(f"\nEvaluating {name} on TEST set (with confusion matrix & report)...")
    test_loss, test_acc = evaluate_model_on_test(model, test_gen, class_indices, name)
    results[name] = {"test_loss": test_loss, "test_acc": test_acc}
    print(f"{name} - Test Accuracy: {test_acc:.4f} | Test Loss: {test_loss:.4f}")

# ============================================================
# FINAL COMPARISON
# ============================================================
print("\n\n================ FINAL TEST RESULTS ================")
for name, res in results.items():
    print(f"{name:22s} -> Acc: {res['test_acc']:.4f} | Loss: {res['test_loss']:.4f}")
print("====================================================")
```

```python
# ===================================================================
# EXPLAINABLE AI-BASED CLASSIFICATION OF BRAIN TUMOR TYPES IN MRI
# HYBRID EfficientNet + ATTENTION + ADVANCED XAI (FASTER TRAINING)
# ===================================================================

# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import os
import random
import warnings

from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import label_binarize
from sklearn.utils.class_weight import compute_class_weight

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    ReduceLROnPlateau,
    ModelCheckpoint,
    EarlyStopping
)
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input as efficientnet_preprocess

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
SEED = 32
tf.random.set_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

print("="*60)
print("BRAIN TUMOR CLASSIFICATION WITH EXPLAINABLE AI")
print("="*60)
print(f"TensorFlow Version: {tf.__version__}")
print(f"GPU Available: {tf.config.list_physical_devices('GPU')}")
print("="*60)

# ===================================================================
# 1. IMPROVED CUSTOM GRAD-CAM IMPLEMENTATION
# ===================================================================

class CustomGradCAM:
    """
    Clean, stable Grad-CAM implementation.
    By default it uses the last conv layer (prefers 'final_conv').
    """

    def __init__(self, model, layer_name=None):
        self.model = model
        self.layer_name = layer_name or self._find_target_layer()

        last_conv_layer = self.model.get_layer(self.layer_name)
        # Model that maps input image to (feature maps, predictions)
        self.grad_model = tf.keras.models.Model(
            inputs=[self.model.inputs],
            outputs=[last_conv_layer.output, self.model.output]
        )
        print(f"[GradCAM] Using layer: {self.layer_name}")

    def _find_target_layer(self):
        # Prefer a layer literally called 'final_conv'
        for layer in self.model.layers:
            if layer.name == "final_conv":
                return "final_conv"

        # Otherwise, fall back to the last Conv2D layer
        for layer in reversed(self.model.layers):
            if isinstance(layer, tf.keras.layers.Conv2D):
                return layer.name

        raise ValueError("No suitable convolutional layer found for Grad-CAM.")

    def compute_heatmap(self, image, class_idx=None, eps=1e-8):
        """
        Compute Grad-CAM heatmap for a batch or single image.

        image: np.array of shape (H, W, 3) or (1, H, W, 3)
               values should be in the SAME SCALE used for training
               (here EfficientNet scale [-1, 1]).
        class_idx: index of the class to explain; if None, uses predicted class.
        Returns: 2D heatmap (float32, 0–1) in feature-map resolution.
        """
        # Ensure batch dimension
        if image.ndim == 3:
            image = np.expand_dims(image, axis=0)

        image = tf.cast(image, tf.float32)

        with tf.GradientTape() as tape:
            conv_outputs, preds = self.grad_model(image)
            if class_idx is None:
                class_idx = tf.argmax(preds[0])
            class_channel = preds[:, class_idx]

        # Gradient of the class score w.r.t. feature maps
        grads = tape.gradient(class_channel, conv_outputs)

        if grads is None:
            # Fallback: uniform weights
            print("[GradCAM] Warning: gradients were None, using uniform weights.")
            weights = tf.ones(conv_outputs.shape[-1])
        else:
            # Global average pooling on gradients
            grads = tf.cast(grads, tf.float32)
            weights = tf.reduce_mean(grads, axis=(0, 1, 2))   # (C,)

        # Weighted sum of feature maps
        conv_outputs = conv_outputs[0]  # (H, W, C)
        cam = tf.tensordot(conv_outputs, weights, axes=([2], [0]))  # (H, W)

        # ReLU activation
        cam = tf.nn.relu(cam)

        # Normalize to [0,1]
        heatmap = cam / (tf.reduce_max(cam) + eps)
        return heatmap.numpy()

# ===================================================================
# 2. DATA ANALYSIS AND LOADING
# ===================================================================

class BrainTumorDataAnalyzer:
    """
    Comprehensive dataset analysis and loading
    """
    
    def __init__(self, base_path):
        self.base_path = base_path
        self.train_path = os.path.join(base_path, 'Training')
        self.test_path = os.path.join(base_path, 'Testing')
        self.classes = self._get_classes()
    
    def _get_classes(self):
        """Get class names from directory structure"""
        if os.path.exists(self.train_path):
            return sorted([d for d in os.listdir(self.train_path) 
                          if os.path.isdir(os.path.join(self.train_path, d))])
        return ['glioma', 'meningioma', 'notumor', 'pituitary']  # Default classes
    
    def analyze_dataset(self):
        """Comprehensive dataset analysis with visualizations"""
        print("\n" + "="*60)
        print("DATASET ANALYSIS")
        print("="*60)
        
        # Count images
        train_counts = {}
        test_counts = {}
        
        for class_name in self.classes:
            train_path = os.path.join(self.train_path, class_name)
            test_path = os.path.join(self.test_path, class_name)
            
            if os.path.exists(train_path):
                train_counts[class_name] = len([
                    f for f in os.listdir(train_path) 
                    if f.lower().endswith(('.jpg', '.png', '.jpeg'))
                ])
            else:
                train_counts[class_name] = 0

            if os.path.exists(test_path):
                test_counts[class_name] = len([
                    f for f in os.listdir(test_path) 
                    if f.lower().endswith(('.jpg', '.png', '.jpeg'))
                ])
            else:
                test_counts[class_name] = 0
        
        # Print statistics
        total_train = sum(train_counts.values())
        total_test = sum(test_counts.values())
        
        print(f"\nDataset Statistics:")
        print(f"Training Set: {total_train} images")
        for class_name, count in train_counts.items():
            pct = (count / total_train * 100) if total_train > 0 else 0
            print(f"  {class_name}: {count} images ({pct:.1f}%)")
        
        print(f"\nTesting Set: {total_test} images")
        for class_name, count in test_counts.items():
            pct = (count / total_test * 100) if total_test > 0 else 0
            print(f"  {class_name}: {count} images ({pct:.1f}%)")
        
        # Simple visualizations
        self._create_visualizations(train_counts, test_counts)
        
        return train_counts, test_counts
    
    def _create_visualizations(self, train_counts, test_counts):
        """Create comprehensive dataset visualizations"""
        fig, axes = plt.subplots(2, 2, figsize=(15, 12))
        
        # Training distribution
        axes[0, 0].bar(train_counts.keys(), train_counts.values(), 
                      color='skyblue', alpha=0.7)
        axes[0, 0].set_title('Training Set Distribution', fontsize=14, fontweight='bold')
        axes[0, 0].set_ylabel('Number of Images')
        axes[0, 0].tick_params(axis='x', rotation=45)
        
        # Testing distribution
        axes[0, 1].bar(test_counts.keys(), test_counts.values(), 
                      color='lightcoral', alpha=0.7)
        axes[0, 1].set_title('Testing Set Distribution', fontsize=14, fontweight='bold')
        axes[0, 1].set_ylabel('Number of Images')
        axes[0, 1].tick_params(axis='x', rotation=45)
        
        # Combined pie chart (train)
        combined_counts = [train_counts.get(cls, 0) for cls in self.classes]
        axes[1, 0].pie(combined_counts, labels=self.classes, autopct='%1.1f%%', 
                       startangle=90, colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99'])
        axes[1, 0].set_title('Overall Class Distribution (Train)', fontsize=14, fontweight='bold')
        
        # Comparison bar chart
        x = np.arange(len(self.classes))
        width = 0.35
        axes[1, 1].bar(x - width/2, [train_counts.get(cls, 0) for cls in self.classes], 
                      width, label='Training', alpha=0.7, color='skyblue')
        axes[1, 1].bar(x + width/2, [test_counts.get(cls, 0) for cls in self.classes], 
                      width, label='Testing', alpha=0.7, color='lightcoral')
        axes[1, 1].set_title('Training vs Testing Distribution', fontsize=14, fontweight='bold')
        axes[1, 1].set_ylabel('Number of Images')
        axes[1, 1].set_xticks(x)
        axes[1, 1].set_xticklabels(self.classes, rotation=45)
        axes[1, 1].legend()
        
        plt.tight_layout()
        plt.show()

# Initialize dataset analyzer
dataset_path = '/kaggle/input/brain-tumor-mri-dataset'
analyzer = BrainTumorDataAnalyzer(dataset_path)
train_counts, test_counts = analyzer.analyze_dataset()
print(f"\nDetected Classes: {analyzer.classes}")

# ===================================================================
# 3. DATA PREPROCESSING + VISUALIZATION
# ===================================================================

class MedicalImagePreprocessor:
    """
    Advanced medical image preprocessing with enhancement techniques,
    then EfficientNet preprocessing.
    """
    
    def __init__(self, img_size=(224, 224), batch_size=32):
        self.img_size = img_size
        self.batch_size = batch_size
    
    def enhance_image(self, img_uint8):
        """
        CLAHE-based enhancement on uint8 RGB image (0-255).
        Returns uint8 RGB 0-255.
        """
        try:
            lab = cv2.cvtColor(img_uint8, cv2.COLOR_RGB2LAB)
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            lab[:, :, 0] = clahe.apply(lab[:, :, 0])
            enhanced = cv2.cvtColor(lab, cv2.COLOR_LAB2RGB)
        except Exception:
            enhanced = img_uint8
        return enhanced

    def preprocess_for_model(self, img):
        """
        Function used by ImageDataGenerator.
        Input img: float32, usually in [0,255].
        Output: float32 preprocessed for EfficientNetB0 ([-1,1]).
        """
        img_uint8 = np.clip(img, 0, 255).astype(np.uint8)
        enhanced_uint8 = self.enhance_image(img_uint8)
        enhanced_float = enhanced_uint8.astype(np.float32)
        enhanced_pre = efficientnet_preprocess(enhanced_float)  # -> [-1, 1]
        return enhanced_pre
    
    def create_data_generators(self, train_path, test_path, validation_split=0.15):
        """
        Create data generators with medical-specific augmentations
        """
        print("\n" + "="*60)
        print("CREATING DATA GENERATORS")
        print("="*60)
        
        # Training data generator with augmentations
        train_datagen = ImageDataGenerator(
            preprocessing_function=self.preprocess_for_model,
            rotation_range=15,
            width_shift_range=0.1,
            height_shift_range=0.1,
            shear_range=0.1,
            zoom_range=0.15,
            horizontal_flip=True,
            vertical_flip=False,
            fill_mode='nearest',
            validation_split=validation_split
        )
        
        # Validation and test generator (no augmentation, only preprocessing)
        test_datagen = ImageDataGenerator(
            preprocessing_function=self.preprocess_for_model
        )
        
        # Create generators
        train_generator = train_datagen.flow_from_directory(
            train_path,
            target_size=self.img_size,
            batch_size=self.batch_size,
            class_mode='categorical',
            subset='training',
            shuffle=True,
            seed=SEED
        )
        
        validation_generator = train_datagen.flow_from_directory(
            train_path,
            target_size=self.img_size,
            batch_size=self.batch_size,
            class_mode='categorical',
            subset='validation',
            shuffle=True,
            seed=SEED
        )
        
        test_generator = test_datagen.flow_from_directory(
            test_path,
            target_size=self.img_size,
            batch_size=self.batch_size,
            class_mode='categorical',
            shuffle=False
        )
        
        print(f"\nData Generator Summary:")
        print(f"Training samples: {train_generator.samples}")
        print(f"Validation samples: {validation_generator.samples}")
        print(f"Testing samples: {test_generator.samples}")
        print(f"Class indices: {train_generator.class_indices}")
        
        return train_generator, validation_generator, test_generator

# ---------- EXTRA VISUALIZATION HELPERS FOR PREPROCESSING & BALANCING ----------

def visualize_preprocessing_examples(train_root, preprocessor, class_names, num_per_class=2):
    """
    Show original vs enhanced images for a few samples per class.
    (Visualization uses 0-1 scale, independent from model preprocessing.)
    """
    print("\n" + "="*60)
    print("VISUALIZING PREPROCESSING (ORIGINAL vs ENHANCED)")
    print("="*60)

    rows = len(class_names)
    cols = num_per_class * 2  # original + enhanced per sample
    plt.figure(figsize=(4 * cols, 3 * rows))

    for r, cls in enumerate(class_names):
        cls_path = os.path.join(train_root, cls)
        if not os.path.exists(cls_path):
            continue
        img_files = [f for f in os.listdir(cls_path)
                     if f.lower().endswith(('.jpg', '.png', '.jpeg'))]
        if len(img_files) == 0:
            continue

        n_show = min(num_per_class, len(img_files))
        chosen = random.sample(img_files, n_show)

        for i, fname in enumerate(chosen):
            img_path = os.path.join(cls_path, fname)
            bgr = cv2.imread(img_path)
            if bgr is None:
                continue
            rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
            rgb_resized = cv2.resize(rgb, (preprocessor.img_size[1],
                                           preprocessor.img_size[0]))
            enhanced_uint8 = preprocessor.enhance_image(rgb_resized.copy())
            enhanced = enhanced_uint8.astype(np.float32) / 255.0
            original = rgb_resized.astype(np.float32) / 255.0

            # Original
            ax = plt.subplot(rows, cols, r * cols + 2 * i + 1)
            ax.imshow(original)
            ax.set_title(f"{cls} - Original", fontsize=9)
            ax.axis('off')

            # Enhanced
            ax = plt.subplot(rows, cols, r * cols + 2 * i + 2)
            ax.imshow(enhanced)
            ax.set_title(f"{cls} - Enhanced", fontsize=9)
            ax.axis('off')

    plt.tight_layout()
    plt.show()

def visualize_batch_from_generator(generator, class_names, title="Augmented Training Batch"):
    """
    Show one batch of images from a generator (what the model actually sees).
    Generator outputs are in EfficientNet scale [-1,1]; we rescale to [0,1] for display.
    """
    print("\n" + "="*60)
    print("VISUALIZING ONE AUGMENTED BATCH")
    print("="*60)
    imgs, labels = next(generator)
    generator.reset()  # reset so training starts from beginning

    n = min(16, imgs.shape[0])
    plt.figure(figsize=(12, 8))
    for i in range(n):
        ax = plt.subplot(4, 4, i + 1)
        img = imgs[i]
        # Map [-1,1] or [0,1] to [0,1] for visualization
        if img.min() < 0:
            img_disp = (img + 1.0) / 2.0
        else:
            img_disp = img
        img_disp = np.clip(img_disp, 0, 1)
        ax.imshow(img_disp)
        cls_idx = np.argmax(labels[i])
        ax.set_title(class_names[cls_idx], fontsize=9)
        ax.axis('off')

    plt.suptitle(title, fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

def compute_and_visualize_class_weights(train_generator):
    """
    Compute class weights for balancing & visualize before/after effect.
    """
    print("\n" + "="*60)
    print("CLASS IMBALANCE ANALYSIS & WEIGHTED BALANCING")
    print("="*60)

    class_indices = train_generator.class_indices  # dict label -> idx
    inv_class_indices = {v: k for k, v in class_indices.items()}

    # train_generator.classes is array of class indices for each sample
    y = train_generator.classes
    classes = np.unique(y)

    # Raw counts
    counts = np.bincount(y)
    class_names_local = [inv_class_indices[i] for i in classes]

    # Compute balanced weights
    weights_arr = compute_class_weight(class_weight='balanced',
                                       classes=classes,
                                       y=y)
    class_weights = {int(c): float(w) for c, w in zip(classes, weights_arr)}

    # Plot raw distribution
    plt.figure(figsize=(14, 5))

    plt.subplot(1, 2, 1)
    sns.barplot(x=class_names_local, y=counts)
    plt.title("Original Training Distribution", fontsize=14, fontweight='bold')
    plt.ylabel("Number of images")
    plt.xticks(rotation=45)

    # "Effective" balanced distribution = count * weight
    balanced_counts = counts * weights_arr

    plt.subplot(1, 2, 2)
    sns.barplot(x=class_names_local, y=balanced_counts)
    plt.title("Effective Distribution After Class Weighting", fontsize=14, fontweight='bold')
    plt.ylabel("Count × Weight")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()

    print("\nComputed class weights (used in model.fit):")
    for idx in classes:
        print(f"  Class {idx} ({inv_class_indices[idx]}): weight = {class_weights[idx]:.4f}")

    return class_weights

# Initialize preprocessor and create generators
preprocessor = MedicalImagePreprocessor(img_size=(224, 224), batch_size=32)
train_gen, val_gen, test_gen = preprocessor.create_data_generators(
    os.path.join(dataset_path, 'Training'),
    os.path.join(dataset_path, 'Testing')
)

# Class names from generator
class_names = list(train_gen.class_indices.keys())

# Visualize preprocessing (original vs enhanced)
visualize_preprocessing_examples(os.path.join(dataset_path, 'Training'),
                                 preprocessor,
                                 class_names,
                                 num_per_class=2)

# Visualize one augmented batch from the training generator
visualize_batch_from_generator(train_gen, class_names)

# Compute class weights and visualize the balancing effect
class_weights = compute_and_visualize_class_weights(train_gen)

# ===================================================================
# 4. ATTENTION MODULES
# ===================================================================

class AttentionModule:
    """
    Custom attention mechanisms for feature refinement
    """
    
    @staticmethod
    def channel_attention(feature_map, ratio=8):
        """Channel Attention Module"""
        channels = feature_map.shape[-1]
        
        # Global Average Pooling branch
        gap = layers.GlobalAveragePooling2D()(feature_map)
        gap = layers.Dense(channels // ratio, activation='relu')(gap)
        gap = layers.Dense(channels, activation='sigmoid')(gap)
        gap = layers.Reshape((1, 1, channels))(gap)
        
        # Global Max Pooling branch
        gmp = layers.GlobalMaxPooling2D()(feature_map)
        gmp = layers.Dense(channels // ratio, activation='relu')(gmp)
        gmp = layers.Dense(channels, activation='sigmoid')(gmp)
        gmp = layers.Reshape((1, 1, channels))(gmp)
        
        # Combine branches
        attention = layers.Add()([gap, gmp])
        attention = layers.Activation('sigmoid')(attention)
        
        # Apply attention
        refined = layers.Multiply()([feature_map, attention])
        
        return refined
    
    @staticmethod
    def spatial_attention(feature_map):
        """Spatial Attention Module"""
        # Average and Max pooling along channel axis
        avg_pool = layers.Lambda(lambda x: tf.reduce_mean(x, axis=3, keepdims=True))(feature_map)
        max_pool = layers.Lambda(lambda x: tf.reduce_max(x, axis=3, keepdims=True))(feature_map)
        
        # Concatenate and apply convolution
        concat = layers.Concatenate(axis=3)([avg_pool, max_pool])
        attention = layers.Conv2D(1, (7, 7), padding='same', activation='sigmoid')(concat)
        
        # Apply attention
        refined = layers.Multiply()([feature_map, attention])
        
        return refined

# ===================================================================
# 5. HYBRID EfficientNet + ATTENTION CLASSIFIER
# ===================================================================

class BrainTumorClassifier:
    """
    Hybrid EfficientNetB0 + Attention CNN architecture
    """
    
    def __init__(self, input_shape=(224, 224, 3), num_classes=4):
        self.input_shape = input_shape
        self.num_classes = num_classes
    
    def build_model(self):
        """
        Build the enhanced CNN model with EfficientNet backbone + attention.
        """
        print("\n" + "="*60)
        print("BUILDING HYBRID EfficientNet + ATTENTION MODEL")
        print("="*60)
        
        # Input layer
        inputs = layers.Input(shape=self.input_shape, name='input_layer')
        
        # EfficientNetB0 backbone
        base_model = EfficientNetB0(
            include_top=False,
            weights='imagenet',
            input_tensor=inputs,
            pooling=None
        )
        
        # Fine-tuning strategy:
        #   - freeze most layers
        #   - unfreeze last ~20 conv layers (except BatchNorm) for better accuracy
        for layer in base_model.layers:
            layer.trainable = False
        for layer in base_model.layers[-20:]:
            if not isinstance(layer, layers.BatchNormalization):
                layer.trainable = True
        
        x = base_model.output   # shape ~ (7, 7, 1280)
        
        # 1×1 conv to reduce channels
        x = layers.Conv2D(512, (1, 1), activation='relu',
                          padding='same', name='reduce_conv')(x)
        x = layers.BatchNormalization(name='reduce_bn')(x)
        
        # Attention block (works at ~7×7)
        x = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='block4_conv1')(x)
        x = layers.BatchNormalization(name='block4_bn1')(x)
        x = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='block4_conv2')(x)
        x = layers.BatchNormalization(name='block4_bn2')(x)
        
        # Channel attention
        x = AttentionModule.channel_attention(x, ratio=8)
        
        x = layers.Conv2D(256, (3, 3), activation='relu', padding='same', name='block4_conv3')(x)
        x = layers.BatchNormalization(name='block4_bn3')(x)
        
        # Spatial attention
        x = AttentionModule.spatial_attention(x)
        
        x = layers.MaxPooling2D((2, 2), name='block4_pool')(x)  # ~3×3
        x = layers.Dropout(0.3, name='block4_dropout')(x)
        
        # Final feature extraction (this is the Grad-CAM target)
        x = layers.Conv2D(512, (3, 3), activation='relu', padding='same', name='final_conv')(x)
        x = layers.BatchNormalization(name='final_bn')(x)
        x = layers.GlobalAveragePooling2D(name='gap')(x)
        x = layers.Dropout(0.4, name='final_dropout')(x)
        
        # Multi-scale feature fusion (dual-branch)
        branch1 = layers.Dense(256, activation='relu',
                              kernel_regularizer=keras.regularizers.l2(0.0005),
                              name='fc_branch1')(x)
        branch1 = layers.BatchNormalization(name='fc_bn1')(branch1)
        branch1 = layers.Dropout(0.3, name='fc_dropout1')(branch1)
        
        branch2 = layers.Dense(128, activation='relu',
                              kernel_regularizer=keras.regularizers.l2(0.0005),
                              name='fc_branch2')(x)
        branch2 = layers.BatchNormalization(name='fc_bn2')(branch2)
        branch2 = layers.Dropout(0.3, name='fc_dropout2')(branch2)
        
        # Concatenate branches
        concatenated = layers.Concatenate(name='feature_concat')([branch1, branch2])
        
        # Final classification layers
        x = layers.Dense(64, activation='relu', name='fc_final')(concatenated)
        x = layers.BatchNormalization(name='fc_final_bn')(x)
        x = layers.Dropout(0.2, name='fc_final_dropout')(x)
        
        # Output layer
        outputs = layers.Dense(self.num_classes, activation='softmax', name='output')(x)
        
        # Create model
        model = keras.Model(inputs=inputs, outputs=outputs, 
                           name='Hybrid_EfficientNet_BrainTumor_Classifier')
        
        return model
    
    def compile_model(self, model, learning_rate=0.0003):
        """
        Compile the model with optimized settings
        """
        optimizer = Adam(learning_rate=learning_rate)
        
        model.compile(
            optimizer=optimizer,
            loss='categorical_crossentropy',
            metrics=[
                keras.metrics.CategoricalAccuracy(name='accuracy'),
                keras.metrics.Precision(name='precision'),
                keras.metrics.Recall(name='recall'),
                keras.metrics.AUC(name='auc')
            ]
        )
        
        return model

# Build and compile the model
num_classes = len(train_gen.class_indices)
classifier = BrainTumorClassifier(input_shape=(224, 224, 3), num_classes=num_classes)
model = classifier.build_model()
model = classifier.compile_model(model, learning_rate=0.0003)

# Display model summary in table form
print("\nModel Architecture:")
model.summary(line_length=120, expand_nested=True)

# Extra: print feature-map design in "C @ HxW" style
def print_feature_map_design(m):
    print("\nFeature-map design for one input image:")
    for layer in m.layers:
        try:
            out_shape = layer.output_shape
        except:
            continue
        # skip multi-output layers
        if isinstance(out_shape, list):
            continue
        if len(out_shape) == 4:
            _, h, w, c = out_shape
            print(f"{layer.name:18s} -> {int(c)} @ {int(h)}x{int(w)}")
        elif len(out_shape) == 2:
            _, d = out_shape
            print(f"{layer.name:18s} -> {int(d)} features")

print_feature_map_design(model)

# ===================================================================
# 6. MODEL TRAINING (EARLY STOPPING + CLASS WEIGHTS)
# ===================================================================

class ModelTrainer:
    """
    Advanced model training with callbacks and monitoring
    """
    
    def __init__(self, model):
        self.model = model
        self.history = None
    
    def get_callbacks(self):
        """
        Create training callbacks:
          - ReduceLROnPlateau
          - ModelCheckpoint (best val_accuracy)
          - EarlyStopping (to avoid wasting time)
        """
        callbacks = [
            ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-7,
                verbose=1
            ),
            ModelCheckpoint(
                'best_brain_tumor_model.h5',
                monitor='val_accuracy',
                save_best_only=True,
                mode='max',
                verbose=1
            ),
            EarlyStopping(
                monitor='val_loss',
                patience=8,
                restore_best_weights=True,
                verbose=1
            )
        ]
        return callbacks
    
    def train(self, train_gen, val_gen, epochs=40, class_weights=None):
        """
        Train the model.
        NOTE: First epoch is always slower because TensorFlow builds the graph.
        """
        print("\n" + "="*60)
        print("STARTING MODEL TRAINING")
        print("="*60)
        print(f"Max epochs: {epochs} (EarlyStopping will stop earlier)")
        print(f"Batch size: {train_gen.batch_size}")
        print(f"Training samples: {train_gen.samples}")
        print(f"Validation samples: {val_gen.samples}")
        print(f"Using class weights: {class_weights is not None}")
        print("="*60)

        steps_per_epoch = train_gen.samples // train_gen.batch_size
        val_steps = max(1, val_gen.samples // val_gen.batch_size)
        
        history = self.model.fit(
            train_gen,
            epochs=epochs,
            validation_data=val_gen,
            steps_per_epoch=steps_per_epoch,
            validation_steps=val_steps,
            callbacks=self.get_callbacks(),
            class_weight=class_weights,
            verbose=1   # <-- removed workers/use_multiprocessing/max_queue_size
        )
        
        self.history = history.history
        return self.history

# Train the model with computed class weights for balancing
trainer = ModelTrainer(model)
history = trainer.train(train_gen, val_gen, epochs=40, class_weights=class_weights)

# Optionally load best checkpoint (based on val_accuracy)
if os.path.exists("best_brain_tumor_model.h5"):
    print("\nLoading best model from checkpoint (highest val_accuracy)...")
    best_model = keras.models.load_model('best_brain_tumor_model.h5', compile=False)
    model = classifier.compile_model(best_model, learning_rate=0.0001)  # lower LR for fine-tuned model

# ===================================================================
# 7. MODEL EVALUATION
# ===================================================================

class ModelEvaluator:
    """
    Comprehensive model evaluation with visualizations
    """
    
    def __init__(self, model, test_generator, class_names):
        self.model = model
        self.test_gen = test_generator
        self.class_names = class_names
        self.y_true = None
        self.y_pred = None
        self.y_pred_proba = None
    
    def evaluate(self):
        """
        Perform comprehensive evaluation
        """
        print("\n" + "="*60)
        print("MODEL EVALUATION")
        print("="*60)
        
        # Get predictions
        self.test_gen.reset()
        self.y_true = self.test_gen.classes
        self.y_pred_proba = self.model.predict(self.test_gen, verbose=1)
        self.y_pred = np.argmax(self.y_pred_proba, axis=1)
        
        # Basic metrics
        test_loss, test_acc, test_prec, test_rec, test_auc = self.model.evaluate(
            self.test_gen, verbose=0
        )
        
        print(f"\nTest Metrics:")
        print(f"  Loss: {test_loss:.4f}")
        print(f"  Accuracy: {test_acc:.4f}")
        print(f"  Precision: {test_prec:.4f}")
        print(f"  Recall: {test_rec:.4f}")
        print(f"  AUC: {test_auc:.4f}")
        
        # Detailed report
        print(f"\nClassification Report:")
        print(classification_report(self.y_true, self.y_pred, 
                                  target_names=self.class_names))
        
        # Visualizations
        self._plot_confusion_matrix()
        self._plot_roc_curves()
        
        return test_acc, test_auc
    
    def _plot_confusion_matrix(self):
        """Plot confusion matrix"""
        cm = confusion_matrix(self.y_true, self.y_pred)
        
        plt.figure(figsize=(10, 8))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=self.class_names,
                   yticklabels=self.class_names,
                   cbar_kws={'shrink': 0.8})
        plt.title('Confusion Matrix', fontsize=16, fontweight='bold')
        plt.ylabel('True Label', fontsize=12)
        plt.xlabel('Predicted Label', fontsize=12)
        plt.tight_layout()
        plt.show()
    
    def _plot_roc_curves(self):
        """Plot ROC curves for each class"""
        y_true_bin = label_binarize(self.y_true, classes=list(range(len(self.class_names))))
        
        fpr = {}
        tpr = {}
        roc_auc_dict = {}
        
        plt.figure(figsize=(10, 8))
        
        for i in range(len(self.class_names)):
            fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], self.y_pred_proba[:, i])
            roc_auc_dict[i] = auc(fpr[i], tpr[i])
            plt.plot(fpr[i], tpr[i], lw=2,
                    label=f'{self.class_names[i]} (AUC = {roc_auc_dict[i]:.3f})')
        
        plt.plot([0, 1], [0, 1], 'k--', lw=2, label='Random Classifier')
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate', fontsize=12)
        plt.ylabel('True Positive Rate', fontsize=12)
        plt.title('Multi-class ROC Curves', fontsize=16, fontweight='bold')
        plt.legend(loc="lower right")
        plt.grid(alpha=0.3)
        plt.tight_layout()
        plt.show()

# Evaluate the model
evaluator = ModelEvaluator(model, test_gen, class_names)
test_accuracy, test_auc = evaluator.evaluate()

# ===================================================================
# 8. TRAINING HISTORY VISUALIZATION
# ===================================================================

def plot_training_history(history):
    """
    Visualize training history
    """
    print("\n" + "="*60)
    print("TRAINING HISTORY VISUALIZATION")
    print("="*60)
    
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    
    metrics = [
        ('accuracy', 'Accuracy'),
        ('loss', 'Loss'),
        ('precision', 'Precision'),
        ('recall', 'Recall'),
        ('auc', 'AUC')
    ]
    
    for idx, (metric, title) in enumerate(metrics):
        row = idx // 3
        col = idx % 3
        
        if metric in history:
            axes[row, col].plot(history[metric], label=f'Training {title}', linewidth=2)
            if f'val_{metric}' in history:
                axes[row, col].plot(history[f'val_{metric}'], label=f'Validation {title}', linewidth=2)
            axes[row, col].set_title(f'Model {title}', fontsize=14, fontweight='bold')
            axes[row, col].set_xlabel('Epoch')
            axes[row, col].set_ylabel(title)
            axes[row, col].legend()
            axes[row, col].grid(alpha=0.3)
    
    # Remove empty subplot if any
    axes[1, 2].axis('off')
    
    plt.tight_layout()
    plt.show()

plot_training_history(history)

# ===================================================================
# 9. COMPREHENSIVE XAI ANALYSIS FOR EACH TUMOR TYPE
# ===================================================================

class TumorSpecificXAI:
    """
    Comprehensive XAI analysis specifically for each tumor type,
    using improved Grad-CAM and stronger focus on the most activated
    tumor regions via thresholding.
    """
    
    def __init__(self, model, class_names, img_size=(224, 224)):
        self.model = model
        self.class_names = class_names
        self.img_size = img_size
        # Default Grad-CAM layer is 'final_conv' if it exists
        self.gradcam_layer_name = "final_conv"
    
    def apply_heatmap(self, image, heatmap, alpha=0.35, threshold=0.4):
        """
        Blend a 2D heatmap with the original image.

        image: model input image (may be in [-1,1] or [0,1]).
        For visualization we map it to [0,1] first.
        """
        # Map to [0,1] for visualization
        if image.min() < 0:
            vis = (image + 1.0) / 2.0
        else:
            vis = image
        vis = np.clip(vis, 0, 1)

        h, w = vis.shape[:2]
        
        # Resize heatmap to image size
        heatmap_resized = cv2.resize(heatmap, (w, h))
        
        # Optional threshold to focus on tumor core region
        if threshold is not None:
            mask = heatmap_resized >= threshold
            if mask.any():
                heatmap_resized = heatmap_resized * mask
                heatmap_resized = heatmap_resized / (heatmap_resized.max() + 1e-8)
        
        heatmap_uint8 = np.uint8(255 * heatmap_resized)
        heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
        
        img_uint8 = np.uint8(np.clip(vis * 255.0, 0, 255))
        img_bgr = cv2.cvtColor(img_uint8, cv2.COLOR_RGB2BGR)
        
        overlay_bgr = cv2.addWeighted(img_bgr, 1 - alpha, heatmap_color, alpha, 0)
        overlay_rgb = cv2.cvtColor(overlay_bgr, cv2.COLOR_BGR2RGB)
        
        return overlay_rgb
    
    def get_samples_by_class(self, test_generator, num_samples=3):
        """
        Get specific number of samples for each class
        """
        test_generator.reset()
        samples_by_class = {class_name: [] for class_name in self.class_names}
        
        # Keep collecting samples until we have enough for each class
        while True:
            batch_images, batch_labels = next(test_generator)
            
            for i in range(len(batch_images)):
                true_label_idx = np.argmax(batch_labels[i])
                class_name = self.class_names[true_label_idx]
                
                if len(samples_by_class[class_name]) < num_samples:
                    samples_by_class[class_name].append({
                        'image': batch_images[i],
                        'true_label': true_label_idx,
                        'class_name': class_name
                    })
            
            # Check if we have enough samples for all classes
            if all(len(samples) >= num_samples for samples in samples_by_class.values()):
                break
            
            # Safety check to prevent infinite loop
            if test_generator.batch_index == 0:
                break
        
        return samples_by_class
    
    def analyze_tumor_type(self, tumor_type, samples, layer_name=None):
        """
        Analyze specific tumor type with detailed XAI.
        """
        print(f"\n{'='*80}")
        print(f"XAI ANALYSIS FOR {tumor_type.upper()} TUMOR TYPE")
        print(f"{'='*80}")
        
        target_layer = layer_name or self.gradcam_layer_name
        gradcam = CustomGradCAM(self.model, layer_name=target_layer)
        
        # Create figure for this tumor type
        fig, axes = plt.subplots(len(samples), 4, figsize=(20, 5*len(samples)))
        if len(samples) == 1:
            axes = axes.reshape(1, -1)
        
        for idx, sample in enumerate(samples):
            img = sample['image']
            true_label = sample['true_label']
            
            # Prepare image
            img_array = np.expand_dims(img, axis=0)
            
            # Get prediction
            prediction = self.model.predict(img_array, verbose=0)[0]
            pred_class = np.argmax(prediction)
            confidence = np.max(prediction)
            
            # Generate Grad-CAM heatmap focused on predicted class
            heatmap = gradcam.compute_heatmap(img_array, class_idx=pred_class)
            
            # Original image
            img_vis = (np.clip((img + 1.0)/2.0, 0, 1) if img.min() < 0 else np.clip(img, 0, 1))
            axes[idx, 0].imshow(img_vis)
            axes[idx, 0].set_title(f'{tumor_type} - Original\nSample {idx+1}', 
                                  fontweight='bold', fontsize=12)
            axes[idx, 0].axis('off')
            
            # Heatmap (upsampled for visualization)
            heatmap_vis = cv2.resize(heatmap, (self.img_size[1], self.img_size[0]))
            axes[idx, 1].imshow(heatmap_vis, cmap='jet')
            axes[idx, 1].set_title('Grad-CAM Heatmap', fontweight='bold', fontsize=12)
            axes[idx, 1].axis('off')
            
            # Overlay
            overlay = self.apply_heatmap(img, heatmap)
            axes[idx, 2].imshow(overlay)
            axes[idx, 2].set_title('Tumor-Region Overlay', fontweight='bold', fontsize=12)
            axes[idx, 2].axis('off')
            
            # Prediction analysis
            axes[idx, 3].axis('off')
            
            # Determine prediction status
            status = "✅ CORRECT" if pred_class == true_label else "❌ INCORRECT"
            status_color = 'green' if pred_class == true_label else 'red'
            
            info_text = f"Prediction Analysis:\n\n"
            info_text += f"True Label: {self.class_names[true_label]}\n"
            info_text += f"Predicted: {self.class_names[pred_class]}\n"
            info_text += f"Confidence: {confidence:.3f}\n"
            info_text += f"Status: {status}\n\n"
            info_text += "Class Probabilities:\n"
            
            for i, (name, prob) in enumerate(zip(self.class_names, prediction)):
                marker = "→" if i == pred_class else " "
                info_text += f"{marker} {name}: {prob:.3f}\n"
            
            axes[idx, 3].text(0.05, 0.95, info_text, transform=axes[idx, 3].transAxes,
                            fontsize=11, verticalalignment='top',
                            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
            
            # Add status indicator
            axes[idx, 3].text(0.95, 0.95, status, transform=axes[idx, 3].transAxes,
                            fontsize=14, fontweight='bold', color=status_color,
                            horizontalalignment='right', verticalalignment='top')
        
        plt.tight_layout()
        plt.show()
        
        # Print summary statistics
        correct_predictions = 0
        for s in samples:
            img_array = np.expand_dims(s['image'], axis=0)
            prediction = self.model.predict(img_array, verbose=0)[0]
            if np.argmax(prediction) == s['true_label']:
                correct_predictions += 1
        accuracy = correct_predictions / len(samples) * 100
        
        print(f"\n{tumor_type} Analysis Summary:")
        print(f"  Samples Analyzed: {len(samples)}")
        print(f"  Correct Predictions: {correct_predictions}")
        print(f"  Accuracy: {accuracy:.1f}%")
    
    def comprehensive_tumor_analysis(self, test_generator, num_samples_per_type=3):
        """
        Perform comprehensive XAI analysis for each tumor type
        """
        print("\n" + "="*80)
        print("COMPREHENSIVE XAI ANALYSIS FOR ALL TUMOR TYPES")
        print("="*80)
        
        # Get samples for each tumor type
        samples_by_class = self.get_samples_by_class(test_generator, num_samples_per_type)
        
        # Analyze each tumor type
        for tumor_type in self.class_names:
            if samples_by_class[tumor_type]:
                self.analyze_tumor_type(tumor_type, samples_by_class[tumor_type])
        
        # Overall comparison
        self.create_comparison_visualization(samples_by_class)
    
    def create_comparison_visualization(self, samples_by_class):
        """
        Create comparison visualization across all tumor types
        """
        print(f"\n{'='*80}")
        print("COMPARATIVE VISUALIZATION ACROSS TUMOR TYPES")
        print(f"{'='*80}")
        
        fig, axes = plt.subplots(len(self.class_names), 3, figsize=(18, 6*len(self.class_names)))
        
        for row_idx, tumor_type in enumerate(self.class_names):
            samples = samples_by_class[tumor_type]
            if not samples:
                continue
            
            # Use first sample for comparison
            sample = samples[0]
            img = sample['image']
            true_label = sample['true_label']
            
            img_array = np.expand_dims(img, axis=0)
            prediction = self.model.predict(img_array, verbose=0)[0]
            pred_class = np.argmax(prediction)
            confidence = np.max(prediction)
            
            gradcam = CustomGradCAM(self.model, layer_name=self.gradcam_layer_name)
            heatmap = gradcam.compute_heatmap(img_array, class_idx=pred_class)
            
            # Original image
            img_vis = (np.clip((img + 1.0)/2.0, 0, 1) if img.min() < 0 else np.clip(img, 0, 1))
            axes[row_idx, 0].imshow(img_vis)
            axes[row_idx, 0].set_title(f'{tumor_type}\nOriginal Image', 
                                       fontweight='bold', fontsize=12)
            axes[row_idx, 0].axis('off')
            
            # Heatmap
            heatmap_vis = cv2.resize(heatmap, (self.img_size[1], self.img_size[0]))
            axes[row_idx, 1].imshow(heatmap_vis, cmap='jet')
            axes[row_idx, 1].set_title('Attention Map', fontweight='bold', fontsize=12)
            axes[row_idx, 1].axis('off')
            
            # Overlay with prediction info
            overlay = self.apply_heatmap(img, heatmap)
            status = "✅" if pred_class == true_label else "❌"
            title = f'{tumor_type}\n{self.class_names[pred_class]} ({confidence:.2f}) {status}'
            axes[row_idx, 2].imshow(overlay)
            axes[row_idx, 2].set_title(title, fontweight='bold', fontsize=12)
            axes[row_idx, 2].axis('off')
        
        plt.tight_layout()
        plt.show()
    
    def generate_attention_statistics(self, test_generator, num_samples=10):
        """
        Generate statistics about attention patterns for each tumor type
        """
        print(f"\n{'='*80}")
        print("ATTENTION PATTERN / PREDICTION STATISTICS")
        print(f"{'='*80}")
        
        samples_by_class = self.get_samples_by_class(test_generator, num_samples)
        
        attention_stats = {}
        
        for tumor_type in self.class_names:
            samples = samples_by_class[tumor_type]
            if not samples:
                continue
            
            confidences = []
            correct_predictions = 0
            
            for sample in samples:
                img_array = np.expand_dims(sample['image'], axis=0)
                prediction = self.model.predict(img_array, verbose=0)[0]
                pred_class = np.argmax(prediction)
                confidence = np.max(prediction)
                
                confidences.append(confidence)
                if pred_class == sample['true_label']:
                    correct_predictions += 1
            
            attention_stats[tumor_type] = {
                'samples': len(samples),
                'correct': correct_predictions,
                'accuracy': correct_predictions / len(samples) * 100,
                'avg_confidence': float(np.mean(confidences)),
                'min_confidence': float(np.min(confidences)),
                'max_confidence': float(np.max(confidences))
            }
        
        # Create visualization
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
        
        # Accuracy comparison
        tumor_types = list(attention_stats.keys())
        accuracies = [attention_stats[t]['accuracy'] for t in tumor_types]
        
        bars = ax1.bar(tumor_types, accuracies,
                       color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'][:len(tumor_types)])
        ax1.set_title('Prediction Accuracy by Tumor Type', fontsize=14, fontweight='bold')
        ax1.set_ylabel('Accuracy (%)')
        ax1.set_ylim(0, 100)
        
        for bar, acc in zip(bars, accuracies):
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height + 1,
                    f'{acc:.1f}%', ha='center', va='bottom', fontweight='bold')
        
        # Confidence comparison
        avg_confidences = [attention_stats[t]['avg_confidence'] for t in tumor_types]
        min_confidences = [attention_stats[t]['min_confidence'] for t in tumor_types]
        max_confidences = [attention_stats[t]['max_confidence'] for t in tumor_types]
        
        x = np.arange(len(tumor_types))
        width = 0.25
        
        ax2.bar(x - width, min_confidences, width, label='Min Confidence', alpha=0.7)
        ax2.bar(x, avg_confidences, width, label='Avg Confidence', alpha=0.7)
        ax2.bar(x + width, max_confidences, width, label='Max Confidence', alpha=0.7)
        
        ax2.set_title('Confidence Distribution by Tumor Type', fontsize=14, fontweight='bold')
        ax2.set_ylabel('Confidence')
        ax2.set_xticks(x)
        ax2.set_xticklabels(tumor_types)
        ax2.legend()
        
        plt.tight_layout()
        plt.show()
        
        # Print detailed statistics
        print("\nDetailed Statistics:")
        for tumor_type in tumor_types:
            stats = attention_stats[tumor_type]
            print(f"\n{tumor_type.upper()}:")
            print(f"  Samples: {stats['samples']}")
            print(f"  Correct Predictions: {stats['correct']}")
            print(f"  Accuracy: {stats['accuracy']:.1f}%")
            print(f"  Average Confidence: {stats['avg_confidence']:.3f}")
            print(f"  Confidence Range: {stats['min_confidence']:.3f} - {stats['max_confidence']:.3f}")

# Perform comprehensive XAI analysis for each tumor type
xai_analyzer = TumorSpecificXAI(model, class_names)
xai_analyzer.comprehensive_tumor_analysis(test_gen, num_samples_per_type=3)
xai_analyzer.generate_attention_statistics(test_gen, num_samples=10)

# ===================================================================
# 10. PERFORMANCE SUMMARY
# ===================================================================

def create_performance_summary(history, test_acc, test_auc, model):
    """
    Create comprehensive performance summary
    """
    print("\n" + "="*80)
    print("RESEARCH PAPER - MODEL PERFORMANCE SUMMARY")
    print("="*80)
    
    # Training metrics
    final_train_acc = history['accuracy'][-1]
    final_val_acc = history['val_accuracy'][-1]
    best_val_acc = max(history['val_accuracy'])
    best_val_auc = max(history['val_auc'])
    actual_epochs = len(history['accuracy'])
    
    print(f"\n📊 PERFORMANCE METRICS:")
    print(f"  Actual Training Epochs:     {actual_epochs}")
    print(f"  Final Training Accuracy:   {final_train_acc:.4f}")
    print(f"  Final Validation Accuracy: {final_val_acc:.4f}")
    print(f"  Best Validation Accuracy:  {best_val_acc:.4f}")
    print(f"  Best Validation AUC:       {best_val_auc:.4f}")
    print(f"  Test Accuracy:             {test_acc:.4f}")
    print(f"  Test AUC:                  {test_auc:.4f}")
    
    # Model info
    total_params = model.count_params()
    trainable_params = sum([tf.keras.backend.count_params(w) 
                           for w in model.trainable_weights])
    
    print(f"\n🏗️ MODEL ARCHITECTURE:")
    print(f"  Total Parameters:      {total_params:,}")
    print(f"  Trainable Parameters:  {trainable_params:,}")
    
    # Novelty contributions
    print(f"\n💡 NOVELTY CONTRIBUTIONS:")
    print(f"  ✅ Hybrid EfficientNetB0 + attention head")
    print(f"  ✅ Medical-specific image enhancement (CLAHE)")
    print(f"  ✅ Multi-scale feature fusion (dual FC branches)")
    print(f"  ✅ Robust Grad-CAM implementation (tumor-focused)")
    print(f"  ✅ Tumor-specific XAI analysis")
    print(f"  ✅ Comparative visualization across tumor types")
    print(f"  ✅ Attention pattern statistics")
    print(f"  ✅ Class-imbalance handling via class weights")
    
    # Clinical relevance
    print(f"\n🏥 CLINICAL RELEVANCE:")
    print(f"  ✓ High accuracy: {test_acc:.1%} (with EfficientNet this can approach ~98% if data supports it)")
    print(f"  ✓ Robust AUC: {test_auc:.3f}")
    print(f"  ✓ Explainable predictions for each tumor type")
    print(f"  ✓ Comparative analysis across tumor types")
    print(f"  ✓ Attention pattern insights")
    print(f"  ✓ Potential for clinical assistance")
    
    print("\n" + "="*80)

# Create performance summary
create_performance_summary(history, test_accuracy, test_auc, model)

# ===================================================================
# 11. SAVE RESULTS
# ===================================================================

# Save model
model.save('enhanced_brain_tumor_classifier.h5')
print("\n✅ Enhanced model saved successfully!")

# Save training history
history_df = pd.DataFrame(history)
history_df.to_csv('enhanced_training_history.csv', index=False)
print("✅ Enhanced training history saved!")

# Save results summary
results = {
    'Test_Accuracy': test_accuracy,
    'Test_AUC': test_auc,
    'Final_Train_Accuracy': history['accuracy'][-1],
    'Final_Val_Accuracy': history['val_accuracy'][-1],
    'Best_Val_Accuracy': max(history['val_accuracy']),
    'Total_Parameters': model.count_params(),
    'Actual_Epochs': len(history['accuracy'])
}

results_df = pd.DataFrame([results])
results_df.to_csv('enhanced_model_results.csv', index=False)
print("✅ Enhanced results summary saved!")

print("\n" + "="*60)
print("🎉 COMPREHENSIVE XAI NOTEBOOK EXECUTION COMPLETED!")
print(f"📊 Model trained for {len(history['accuracy'])} epochs (with early stopping)")
print("📄 All outputs ready for research paper!")
print("🔍 Tumor-specific XAI analysis completed!")
print("📈 Attention pattern statistics generated!")
print("="*60)
```
